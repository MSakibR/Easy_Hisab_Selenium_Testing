```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time
import random

# --- Setup Chrome Driver ---
chrome_options = Options()
chrome_options.add_argument("--start-maximized")
chrome_options.add_experimental_option("detach", True)
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()), options=chrome_options
)
wait = WebDriverWait(driver, 15)

# --- Signup data ---
first_name = "M. Sakib"
last_name = "Rahman"
email = f"sakib{random.randint(1000,9999)}@gmail.com"  # unique email
phone_number = "01712345678"
company_name = "EasyHisab"
password = "Password123"

try:
    # 1️⃣ Go to Signup page
    driver.get("http://localhost:3000/auth")
    time.sleep(2)

    # Click Sign Up tab if needed
    signup_tab = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//button[text()='Sign Up']"))
    )
    signup_tab.click()
    time.sleep(1)

    # 2️⃣ Fill Signup Form
    driver.find_element(By.XPATH, "//input[@placeholder='First Name']").send_keys(
        first_name
    )
    driver.find_element(By.XPATH, "//input[@placeholder='Last Name']").send_keys(
        last_name
    )
    driver.find_element(By.XPATH, "//input[@placeholder='Email']").send_keys(email)
    driver.find_element(By.XPATH, "//input[@placeholder='Phone Number']").send_keys(
        phone_number
    )
    driver.find_element(By.XPATH, "//input[@placeholder='Company Name']").send_keys(
        company_name
    )
    driver.find_element(By.XPATH, "//input[@placeholder='Password']").send_keys(
        password
    )
    driver.find_element(By.XPATH, "//input[@placeholder='Confirm Password']").send_keys(
        password
    )

    # Click Terms checkbox
    driver.find_element(By.XPATH, "//input[@type='checkbox']").click()

    # Submit signup
    driver.find_element(By.XPATH, "//button[text()='Create Account']").click()
    print(f"✅ Signed up with email: {email}")

    # 3️⃣ Wait for Dashboard page load
    wait.until(EC.url_contains("/dashboard"))
    print("✅ Redirected to dashboard after signup")
    time.sleep(2)

    # 4️⃣ Navigate back to Auth page
    driver.get("http://localhost:3000/auth")
    wait.until(EC.url_contains("/auth"))
    print("✅ Returned to auth page")
    time.sleep(2)

    # 5️⃣ Fill Login Form with same info
    login_tab = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//button[text()='Login']"))
    )
    login_tab.click()
    time.sleep(1)

    driver.find_element(By.XPATH, "//input[@type='email']").send_keys(email)
    driver.find_element(By.XPATH, "//input[@type='password']").send_keys(password)

    driver.find_element(By.XPATH, "//button[text()='Sign In']").click()
    print("✅ Logged in with signup credentials")

    
    wait.until(EC.url_contains("/dashboard"))
    print("✅ Successfully logged in again and reached dashboard")

    time.sleep(3)

finally:
    driver.quit()
    print("✅ Browser closed")
```
