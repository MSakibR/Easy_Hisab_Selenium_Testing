```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait, Select
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time

# --- Setup Chrome Driver ---
chrome_options = Options()
chrome_options.add_argument("--start-maximized")
chrome_options.add_experimental_option("detach", True)
prefs = {"credentials_enable_service": False, "profile.password_manager_enabled": False}
chrome_options.add_experimental_option("prefs", prefs)

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()), options=chrome_options
)
wait = WebDriverWait(driver, 15)

try:
    # --- 1️⃣ Go to Login Page ---
    driver.get("http://localhost:3000/auth")

    # --- 2️⃣ Login ---
    email_input = wait.until(
        EC.presence_of_element_located((By.XPATH, "//input[@type='email']"))
    )
    password_input = driver.find_element(By.XPATH, "//input[@type='password']")
    email_input.send_keys("sakib2333@gmail.com")
    password_input.send_keys("Password123")

    sign_in_button = driver.find_element(
        By.XPATH, "//button[contains(text(), 'Sign In')]"
    )
    sign_in_button.click()

    # --- 3️⃣ Wait for Dashboard ---
    wait.until(EC.url_contains("/dashboard"))
    print("✅ Logged in successfully!")
    time.sleep(2)

    # --- 4️⃣ Navigate to Payments Page ---
    driver.get("http://localhost:3000/payment")
    wait.until(EC.url_contains("/payment"))
    print("✅ Payments page loaded!")
    time.sleep(2)

    # --- 5️⃣ Enter Invoice ID in Search Bar ---
    search_input = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//input[@placeholder='Search by Invoice ID or Contact Person']")
        )
    )
    search_input.clear()
    search_input.send_keys("INV-73955")
    print("🔍 Typing search query...")

    # --- 6️⃣ Click Search Button ---
    search_button = driver.find_element(
        By.XPATH, "//button[contains(text(), 'Search')]"
    )
    search_button.click()
    print("▶️ Search button clicked")
    time.sleep(2)  # wait for search results

    # --- 7️⃣ Highlight first row of search results ---
    result_row = wait.until(EC.presence_of_element_located((By.XPATH, "//tbody/tr")))
    driver.execute_script(
        "arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});"
        "arguments[0].style.border='3px solid red';"
        "arguments[0].style.backgroundColor='#fff8dc';",
        result_row,
    )
    print("✨ First search result row highlighted")
    time.sleep(4)  # keep highlight visible

    # --- 8️⃣ Remove highlight BEFORE clicking Cancel ---
    driver.execute_script(
        "arguments[0].style.border=''; arguments[0].style.backgroundColor='';",
        result_row,
    )
    print("🧹 Highlight removed")

    # --- 9️⃣ Click Cancel Button ---
    cancel_button = driver.find_element(
        By.XPATH, "//button[contains(text(), 'Cancel')]"
    )
    cancel_button.click()
    print("✅ Cancel clicked to clear search")
    time.sleep(2)

finally:
    driver.quit()
    print("✅ Browser closed!")
```
