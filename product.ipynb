```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time


# ✅ Highlight helper function (for buttons, modals)
def highlight(driver, element, duration=2):
    driver.execute_script(
        "arguments[0].setAttribute('style', arguments[0].getAttribute('style') + ';"
        "border: 3px solid red; background-color: yellow;');",
        element,
    )
    time.sleep(duration)
    driver.execute_script(
        "arguments[0].setAttribute('style', arguments[0].getAttribute('style').replace("
        "'border: 3px solid red; background-color: yellow;', ''));",
        element,
    )


# ✅ Highlight table row function (for added/updated rows)
def highlight_row(driver, xpath, duration=2, color="lightgreen"):
    row = driver.find_element(By.XPATH, xpath)
    driver.execute_script(f"arguments[0].style.backgroundColor = '{color}';", row)
    time.sleep(duration)
    driver.execute_script("arguments[0].style.backgroundColor = '';", row)


# ✅ Setup Chrome
chrome_options = Options()
chrome_options.add_argument("--start-maximized")
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()), options=chrome_options
)

wait = WebDriverWait(driver, 10)

try:
    # --- LOGIN ---
    print("🔐 Logging in...")
    driver.get("http://localhost:3000/auth")  # login page
    email_input = wait.until(
        EC.presence_of_element_located((By.XPATH, "//input[@type='email']"))
    )
    password_input = driver.find_element(By.XPATH, "//input[@type='password']")
    email_input.send_keys("sakib2333@gmail.com")  # your email
    password_input.send_keys("Password123")  # your password

    sign_in_button = driver.find_element(
        By.XPATH, "//button[contains(text(), 'Sign In')]"
    )
    
    sign_in_button.click()

    # wait until dashboard loads
    wait.until(EC.url_contains("/dashboard"))
    print("✅ Logged in successfully! Navigated to dashboard.")
    time.sleep(3)

    # --- NAVIGATE TO PRODUCTS PAGE ---
    print("🧭 Navigating to Product page...")
    driver.get("http://localhost:3000/product")  # adjust route if needed
    wait.until(
        EC.presence_of_element_located((By.XPATH, "//h1[contains(text(), 'Products')]"))
    )
    time.sleep(3)

    # --- ADD PRODUCT ---
    print("➕ Opening Add Product modal...")
    add_btn = wait.until(
        EC.element_to_be_clickable(
            (By.XPATH, "//button[contains(text(), '+ Add Product')]")
        )
    )
    
    add_btn.click()
    time.sleep(3)

    print("💾 Filling Add Product form...")
    wait.until(
        EC.presence_of_element_located((By.XPATH, "//input[@placeholder='Name']"))
    ).send_keys("Test Product 2025")
    driver.find_element(By.XPATH, "//input[@placeholder='Category']").send_keys(
        "Electronics"
    )
    driver.find_element(By.XPATH, "//input[@placeholder='Unit Price']").send_keys(
        "2500"
    )
    driver.find_element(By.XPATH, "//input[@placeholder='Stock Quantity']").send_keys(
        "25"
    )

    save_btn = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Save')]"))
    )
    
    save_btn.click()
    print("✅ Product added successfully!")
    time.sleep(3)

    # Highlight newly added row
    highlight_row(
        driver,
        "//tr[td[contains(text(), 'Test Product 2025')]]",
        duration=3,
        color="lightblue",
    )

    # --- VIEW PRODUCT ---
    print("👁️ Viewing Product Details...")
    view_btn = wait.until(
        EC.element_to_be_clickable(
            (
                By.XPATH,
                "//tr[td[contains(text(), 'Test Product 2025')]]//button[contains(text(), 'View')]",
            )
        )
    )
    
    view_btn.click()
    time.sleep(3)

    close_view_btn = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Close')]"))
    )
    close_view_btn.click()
    print("🚪 Closed View Modal.")
    time.sleep(3)

    # --- EDIT PRODUCT ---
    print("✏️ Opening Edit Modal...")
    edit_btn = wait.until(
        EC.element_to_be_clickable(
            (
                By.XPATH,
                "//tr[td[contains(text(), 'Test Product 2025')]]//button[contains(text(), 'Edit')]",
            )
        )
    )
    
    edit_btn.click()
    time.sleep(3)

    # Change only Name (not price)
    name_field = wait.until(
        EC.presence_of_element_located((By.XPATH, "//input[@placeholder='Name']"))
    )
    name_field.clear()
    name_field.send_keys("Updated Product 2025")

    update_btn = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Update')]"))
    )
    
    update_btn.click()
    print("✅ Product updated successfully!")
    time.sleep(3)

    # Highlight updated row
    highlight_row(
        driver,
        "//tr[td[contains(text(), 'Updated Product 2025')]]",
        duration=3,
        color="lightgreen",
    )

    # --- DELETE PRODUCT ---
    print("🗑️ Deleting Product...")
    delete_btn = wait.until(
        EC.element_to_be_clickable(
            (
                By.XPATH,
                "//tr[td[contains(text(), 'Updated Product 2025')]]//button[contains(text(), 'Delete')]",
            )
        )
    )
    
    delete_btn.click()
    print("✅ Product deleted successfully!")
    time.sleep(3)

except Exception as e:
    print(f"⚠️ Error: {e}")

finally:
    print("✅ Test completed. Closing browser in 3 seconds...")
    time.sleep(3)
    driver.quit()
```
