```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time
import random

# --- Setup Chrome Driver ---
chrome_options = Options()
chrome_options.add_argument("--start-maximized")
chrome_options.add_experimental_option("detach", True)
prefs = {"credentials_enable_service": False, "profile.password_manager_enabled": False}
chrome_options.add_experimental_option("prefs", prefs)

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()), options=chrome_options
)
wait = WebDriverWait(driver, 15)

try:
    # --- 1️⃣ Go to Login Page ---
    driver.get("http://localhost:3000/auth")

    # --- 2️⃣ Login ---
    email_input = wait.until(
        EC.presence_of_element_located((By.XPATH, "//input[@type='email']"))
    )
    password_input = driver.find_element(By.XPATH, "//input[@type='password']")
    email_input.send_keys("sakib2333@gmail.com")
    password_input.send_keys("Password123")

    sign_in_button = driver.find_element(
        By.XPATH, "//button[contains(text(), 'Sign In')]"
    )
    sign_in_button.click()

    # --- 3️⃣ Wait for Dashboard ---
    wait.until(EC.url_contains("/dashboard"))
    print("✅ Logged in successfully!")

    time.sleep(2)

    # --- 4️⃣ Navigate to Help Page ---
    driver.get("http://localhost:3000/help")
    wait.until(EC.url_contains("/help"))
    print("✅ Help page loaded!")
    time.sleep(2)

    # --- Smooth Scroll Function ---
    def smooth_scroll(y_distance):
        driver.execute_script(
            """
            window.scrollBy({top: arguments[0], left: 0, behavior: 'smooth'});
        """,
            y_distance,
        )
        time.sleep(1.5)

    # --- Scroll down in steps ---
    for _ in range(5):
        smooth_scroll(300)

    # --- 5️⃣ Open Chat Widget ---
    chat_button = wait.until(
        EC.element_to_be_clickable(
            (By.XPATH, "//button[contains(text(), '💬 Chat Support')]")
        )
    )
    chat_button.click()
    time.sleep(2)

    # --- 6️⃣ Close Chat Widget ---
    close_chat = wait.until(
        EC.element_to_be_clickable(
            (By.XPATH, "//div[contains(@class,'fixed')]//button[text()='×']")
        )
    )
    close_chat.click()
    time.sleep(1)

    # --- 7️⃣ Scroll to FAQ Section ---
    faq_section = driver.find_element(By.ID, "faq-section")
    driver.execute_script(
        "arguments[0].scrollIntoView({behavior: 'smooth'});", faq_section
    )
    time.sleep(2)

    # --- 8️⃣ Expand Random 2-3 FAQ Items ---
    faq_buttons = driver.find_elements(By.XPATH, "//section[@id='faq-section']//button")
    random_faqs = random.sample(faq_buttons, k=min(3, len(faq_buttons)))
    for btn in random_faqs:
        driver.execute_script(
            "arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});", btn
        )
        time.sleep(1)
        btn.click()
        time.sleep(1)

    print("✅ Help page actions completed!")

finally:
    time.sleep(5)
    driver.quit()
```
