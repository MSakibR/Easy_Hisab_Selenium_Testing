```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait, Select
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time

# --- Setup Chrome Driver ---
chrome_options = Options()
chrome_options.add_argument("--start-maximized")
chrome_options.add_experimental_option("detach", True)
prefs = {"credentials_enable_service": False, "profile.password_manager_enabled": False}
chrome_options.add_experimental_option("prefs", prefs)

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()), options=chrome_options
)
wait = WebDriverWait(driver, 15)

try:
    # --- 1️⃣ Login ---
    driver.get("http://localhost:3000/auth")
    email_input = wait.until(
        EC.presence_of_element_located((By.XPATH, "//input[@type='email']"))
    )
    password_input = driver.find_element(By.XPATH, "//input[@type='password']")
    email_input.send_keys("sakib2333@gmail.com")
    password_input.send_keys("Password123")
    driver.find_element(By.XPATH, "//button[contains(text(), 'Sign In')]").click()
    wait.until(EC.url_contains("/dashboard"))
    print("✅ Logged in successfully!")
    time.sleep(2)

    # --- 2️⃣ Go to Invoice Management page ---
    driver.get("http://localhost:3000/invoice")
    wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//h1[contains(text(), 'Invoice Management')]")
        )
    )
    print("✅ Invoice page loaded")
    time.sleep(1)

    # --- 3️⃣ Search Invoices ---
    search_input = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//input[@placeholder='Search by invoice or customer...']")
        )
    )
    search_input.clear()
    search_input.send_keys("INV-97876")
    print("🔍 Typing search query...")

    search_button = driver.find_element(
        By.XPATH,
        "//form[.//input[@placeholder='Search by invoice or customer...']]//button[@type='submit']",
    )
    search_button.click()
    print("▶️ Search button clicked")
    time.sleep(2)

    # --- 4️⃣ Highlight the first search result row ---
    result_row = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//tbody/tr[contains(@class, 'hover:bg-gray-50')]")
        )
    )
    driver.execute_script(
        "arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});"
        "arguments[0].style.border='3px solid red'; arguments[0].style.backgroundColor='#fff8dc';",
        result_row,
    )
    print("✨ First search result highlighted")
    time.sleep(3)

    # --- 5️⃣ Cancel search ---
    cancel_button = driver.find_element(
        By.XPATH, "//button[contains(text(), 'Cancel')]"
    )
    cancel_button.click()
    print("🧹 Search canceled")
    time.sleep(2)

    # Remove search highlight
    driver.execute_script(
        "arguments[0].style.border=''; arguments[0].style.backgroundColor='';",
        result_row,
    )

    # --- 6️⃣ Select first unpaid invoice ---
    unpaid_select = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//form[.//button[contains(text(),'Mark as Paid')]]//select")
        )
    )
    select = Select(unpaid_select)
    if len(select.options) > 1:
        select.select_by_index(1)
        print("✅ Selected first unpaid invoice")
    else:
        print("⚠️ No unpaid invoices available")
        raise Exception("No unpaid invoices found")

    # --- 7️⃣ Highlight first unpaid invoice row ---
    unpaid_row = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//form[.//button[contains(text(),'Mark as Paid')]]//tr")
        )
    )
    driver.execute_script(
        "arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});"
        "arguments[0].style.border='3px solid green'; arguments[0].style.backgroundColor='#e6ffe6';",
        unpaid_row,
    )
    print("✨ First unpaid invoice highlighted")
    time.sleep(3)

    # --- 8️⃣ Click Mark as Paid ---
    mark_paid_button = driver.find_element(
        By.XPATH,
        "//form[.//button[contains(text(),'Mark as Paid')]]//button[contains(text(),'Mark as Paid')]",
    )
    mark_paid_button.click()
    print("🚀 Marked invoice as paid")
    time.sleep(3)

    # Remove unpaid highlight
    driver.execute_script(
        "arguments[0].style.border=''; arguments[0].style.backgroundColor='';",
        unpaid_row,
    )

finally:
    driver.quit()
    print("✅ Browser closed!")
```
